```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
image = cv2.imread('two.jpg', cv2.IMREAD_GRAYSCALE)
edges = cv2.Canny(image, 50, 150)
lines = cv2.HoughLinesP(edges, 1, np.pi/180, threshold=100, minLineLength=50,
maxLineGap=10)
line_image = cv2.cvtColor(edges, cv2.COLOR_GRAY2BGR)
for line in lines:
    x1, y1, x2, y2 = line
    cv2.line(line_image, (x1, y1), (x2, y2), (0, 255, 0), 2)
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1), plt.imshow(image, cmap='gray'), plt.title('Original Image')
plt.subplot(1, 2, 2), plt.imshow(line_image), plt.title('Detected Lines')
plt.show()
```
